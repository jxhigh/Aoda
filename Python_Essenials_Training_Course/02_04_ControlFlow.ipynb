## Control Flow

### If / Else statements

```python
a = True
if a:
    print('It is true!')
    print('Also print this')
else:
    print('It is false!')
print('Always print this')
# if a is True... the colon followed by an indent means that those indented belongs to the
# text before the colon- recognized as a 'block' of code
```

```python
a = True
b = False
c = True
if a:
    print('It is true!')
    print('Also print this')
    if b:
        print('Both are true')
        if c:
            print('All three are true')
else:
    print('It is false!')
print('Always print this')
# along the way in the statement if any return as False you never will get those indented
# below it
```

### For loops

```python
a = [1,2,3,4,5]
for number in a:
    print(number)
# loops were made to avoid all the excessive indentations. foor loops loop over iterables like
# a list
```

### While loops

```python
a = 0
while a < 5:
    print(a)
    a = a + 1
```
