## Sets

```python
# A set is almost identical to a list except that all of the elements in it have to be 
    # unique, and a set is indicated by the curly brackets.
my_set = {1,2,3,4,5}
print(my_set)
```

```python
type(my_set)
```

```python
len(my_set)
```

```python
my_set = {1,1,2,2}
len(my_set) # This will show 2 elements bc it will discard those that are not unique like
            # the second 1 and 2s.
```

```python
print(my_set)
```

```python
[1,2] == [2,1] # The order of these elements in a list are different so its false
```

```python
{1,2,3} == {3,2,1,1,1} # The order of these elements in a set doesn't matter that they
                       # are different so its true
```

## Tuples

```python
my_tuple = (1,2,3) #pronounced "two-pulls" and are declared with paretheses
# Tuples are very similiar to lists. The orders matter. The difference is that you cannot
# append or add things to tuples unlike lists.
```

```python
len(my_tuple)
```

```python
(1,2) == (2,1)
```

```python
my_list.append(6)
print(my_list)
```

```python
my_tuple.append(4) # cant modify a tuple after its declared
# Why would you use a tuple? They are memory effecient (python not alloting more memory
# to it than it needs like it does for lists.)
```

## Dictionaries

```python
my_dictionary = {
    'apple': 'A red fruit', 
    'bear': 'A scary animal'
}
# dictionaries are declared with curly brackets like sets. They have these key:value pairs
# the keys have to be unique. Like sets the order doesn't matter either.
```

```python
my_dictionary['apple']
```

```python
my_dictionary = {
    'apple': 'A red fruit', 
    'bear': 'A scary animal',
    'apple': 'Sometimes a green fruit'
}
```

```python
my_dictionary['apple'] # just going to take the second definition
```
